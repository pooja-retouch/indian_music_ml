# Spotify Top 10 Tracks - A.R. Rahman

```python
# ## 1. Setup
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import pandas as pd
import json

# Your Spotify API credentials
CLIENT_ID = "d907d7da04e14ca4886a05e7101557cd"
CLIENT_SECRET = "92f010e221c942ce8383060c29c03cfe"

# Authenticate
sp = spotipy.Spotify(
    auth_manager=SpotifyClientCredentials(
        client_id=CLIENT_ID,
        client_secret=CLIENT_SECRET
    )
)
```

```python
# ## 2. Fetch Top 10 Tracks for A.R. Rahman
# A.R. Rahman Spotify Artist ID: 1mYsTxnqsietFxj1OgoGbG
artist_uri = "spotify:artist:1mYsTxnqsietFxj1OgoGbG"

results = sp.artist_top_tracks(artist_uri, country="IN")
top_tracks = results['tracks'][:10]

# Save raw data for reference
with open("ar_rahman_top_tracks.json", "w", encoding="utf-8") as f:
    json.dump(top_tracks, f, indent=2, ensure_ascii=False)

# Print songs with Spotify URLs
for t in top_tracks:
    print(f"Song: {t['name']}")
    print(f"Spotify URL : {t['external_urls']['spotify']}\n")
```

```python
# ## 3. Convert to DataFrame and Save CSV
songs = [
    {
        "name": t['name'],
        "album": t['album']['name'],
        "release_date": t['album']['release_date'],
        "popularity": t['popularity'],
        "url": t['external_urls']['spotify'],
        "id": t['id']
    }
    for t in top_tracks
]

df = pd.DataFrame(songs)

# Save CSV
df.to_csv("ar_rahman_top10.csv", index=False, encoding="utf-8")

print("✅ Top 10 tracks saved to ar_rahman_top10.csv")
df
```

```python
most_popular = df.loc[df['popularity'].idxmax()]
print(f"🎵 Most Popular Track: {most_popular['name']} (Popularity {most_popular['popularity']})")
```

```python
import matplotlib.pyplot as plt

# Sort by popularity for better visualization
df_sorted = df.sort_values(by="popularity", ascending=True)

plt.figure(figsize=(10,6))
plt.barh(df_sorted["name"], df_sorted["popularity"], color="skyblue")
plt.xlabel("Popularity")
plt.title("A.R. Rahman - Top 10 Tracks Popularity")
plt.tight_layout()
plt.show()
```
